```python
 #3.Find Even and Odd Numbers in a List
#Question: Write a Python program to separate even and odd numbers from the list [12, 17, 19, 24, 35, 40, 56].

EvenNumbers = []
oddNumbers = []
list=[12, 17, 19, 24, 35, 40, 56]
for i in list:
    if i%2==0:
        EvenNumbers.append(i)
    else:
        oddNumbers.append(i)
```
